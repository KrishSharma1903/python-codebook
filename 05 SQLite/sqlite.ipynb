SQL And SQLite
SQL (Structured Query Language) is a standard language for managing and manipulating relational databases. SQLite is a self-contained, serverless, and zero-configuration database engine that is widely used for embedded database systems. 

## CRUD OPERATIONS IN SQLITE

```python
import sqlite3
```

```python
##Connecting to an SQLite database
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
##Create a table
cursor.execute('''
Create Table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text
)
''')

##Commit the changes 
connection.commit()
```

```python
##Create a table
cursor.execute('''
Select * from employees
''')
```

```python
##Insterting the data in the sqlite table 
cursor.execute('''
INSERT INTo employees(name,age,department)
VALUES('Krish',21,'AI')
''')
cursor.execute('''
INSERT INTO employees (name, age, department)
VALUES ('Bob', 25, 'Engineering')
''')

cursor.execute('''
INSERT INTO employees (name, age, department)
VALUES ('Charlie', 35, 'Finance')
''')

## commit the changes
connection.commit()
```

```python
##Query the data from the table 
cursor.execute('Select * FROM employees')
rows= cursor.fetchall()

##print the query data
for row in rows:
    print(row)
```

```python
##Update the data into the table 
cursor.execute('''
UPDATE employees 
set age =22
where name="Krish" ''')

connection.commit()
```

```python
##Query the data from the table 
cursor.execute('Select * FROM employees')
rows= cursor.fetchall()

##print the query data
for row in rows:
    print(row)
```

```python
##Delete the data from the file 
cursor.execute(''' 
Delete from employees 
    where name = 'Bob'
''')

connection.commit()
```
